```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns


def plot_strong_correlations(data, threshold=0.7):
    """Computes and plots a heatmap of numerical features with a correlation

    magnitude greater than or equal to the specified threshold.
    """
    # 1. Isolate the numerical columns
    df_numeric = data.select_dtypes(include=["number"])

    # Edge case check: Ensure there are actual numerical columns to plot
    if df_numeric.empty:
        print("Error: The provided DataFrame contains no numerical columns.")
        return

    # 2. Compute the Pearson correlation matrix
    corr_matrix = df_numeric.corr()

    # 3. Filter for strong correlations based on the threshold parameter
    strong_corr = corr_matrix[corr_matrix.abs() >= threshold]

    # 4. Initialize the plot layout using subplots
    fig, ax = plt.subplots(figsize=(14, 12))

    # 5. Draw the correlation heatmap
    sns.heatmap(
        strong_corr,
        annot=True,
        cmap="coolwarm",
        fmt=".2f",
        linewidths=0.1,
        vmax=1,
        vmin=-1,
        ax=ax,
    )

    # 6. Style the axes and layout dynamically based on the threshold value
    ax.set_title(
        f"Strong Feature Correlation Heatmap ($|r| \geq {threshold}$)",
        fontsize=16,
        pad=20,
    )
    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)
    plt.tight_layout()

    # 7. Save and display the image
    filename = f"strong_correlation_heatmap_{threshold}.png"
    plt.savefig(filename, dpi=300, bbox_inches="tight")
    plt.show()
```
